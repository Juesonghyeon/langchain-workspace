```python
from typing import Annotated
from typing_extensions import TypedDict 
from langgraph.graph.message import add_messages

class State(TypedDict):
    messages: Annotated[list[str], add_messages]
```

```python
def check_weather_keyword(state: State) -> str:
    """사용자 메시지에 "날씨" 키워드가 있는지 확인하는 노드"""
    last_message = state["messages"][-1]

    if "날씨" in last_message.content:
        print("날씨 키워드 감지!")
        return "weather_node"
    else:
        print("일반 키워드 감지!")
        return "general_node"
```

```python
def provide_weather(state: State) -> State:
    """날씨 정보를 제공하는 노드"""
    print("날씨 노드 실행: 날씨 정보를 제공합니다.")
    return {"messages": ["오늘 날씨는 맑고 따뜻합니다."]}
```

```python
def provide_general(state: State) -> State:
    """일반적인 답변을 제공하는 노드"""
    print("일반 노드 실행: 일반적인 답변을 제공합니다.")
    return {"messages": ["일반적인 답변입니다."]}
```

```python
from langgraph.graph import StateGraph

graph_builder = StateGraph(State)
```

```python
graph_builder.add_node("weather_node", provide_weather)
graph_builder.add_node("general_node", provide_general)
```

```python
from langgraph.graph import START, END

# 조건부 엣지
graph_builder.add_conditional_edges(
    START,
    check_weather_keyword,
    {
        "weather_node": "weather_node",
        "general_node": "general_node"
    }
)
```

```python
graph_builder.add_edge("weather_node", END)
graph_builder.add_edge("general_node", END)
```

```python
graph = graph_builder.compile()
```

```python
from IPython.display import Image, display

display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
graph.invoke({"messages": ["오늘 뭐 먹지?"]})
```
